```python
import os
from getpass import getpass
from pathlib import Path

from IPython.display import Image, Video, display
from ultralytics import YOLO
from ultralytics.data.converter import convert_ndjson_to_yolo

os.environ["ULTRALYTICS_API_KEY"] = getpass("API key de Ultralytics: ")
DATASET = await convert_ndjson_to_yolo("https://platform.ultralytics.com/lucas-estrada/datasets/super-vivencia")
print(f"Dataset listo en: {DATASET}")
```

```python
model = YOLO("yolo26n.pt")
model.train(data=DATASET, epochs=100, imgsz=640, batch=-1, name="super-vivencia")

carpeta = Path(model.trainer.save_dir)
BEST = carpeta / "weights" / "best.pt"
print(f"Pesos del mejor modelo: {BEST}")
```
